```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/tips.csv')
```

```python
df.head()
```

```python
print('Средний счёт:', round(df['total_bill'].mean(), 2))
print('Средние чаевые:', round(df['tip'].mean(), 2))
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['total_bill'], df['tip'], alpha=0.6)
plt.xlabel('Сумма счёта')
plt.ylabel('Чаевые')
plt.title('Зависимость чаевых от суммы счёта')
plt.grid(True)
plt.show()
```
